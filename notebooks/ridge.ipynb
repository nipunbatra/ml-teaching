```python
#Define input array with angles from 60deg to 300deg converted to radians
x = np.array([i*np.pi/180 for i in range(60,300,4)])
np.random.seed(10)  #Setting seed for reproducability
y = 4*x + 7 + np.random.normal(0,3,len(x))
y_true = 4*x + 7
max_deg = 20
data_x = [x**(i+1) for i in range(max_deg)] + [y]
data_c = ['x'] + ['x_{}'.format(i+1) for i in range(1,max_deg)] + ['y']
data = pd.DataFrame(np.column_stack(data_x),columns=data_c)
data["ones"] = 1
plt.plot(data['x'],data['y'],'.', label='Data Points')
plt.plot(data['x'], y_true,'g', label='True Function')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
format_axes(plt.gca())
plt.savefig('lin_1.pdf', transparent=True, bbox_inches="tight")
```

```python
from sklearn.preprocessing import PolynomialFeatures
for i, deg in enumerate([1, 2, 3, 4, 5, 6, 7, 8, 10, 12, 15, 20, 25]):
    poly = PolynomialFeatures(degree=deg, include_bias=False)
    X_ = poly.fit_transform(x.reshape(-1, 1))
    from sklearn.linear_model import LinearRegression
    clf = LinearRegression()
    clf.fit(X_, data['y'])
    print(X_.shape)
    y_pred = clf.predict(X_)
    plt.figure()
    plt.plot(data['x'], y_pred, label='Degree: {}'.format(deg))
    plt.xlabel("x")
    plt.ylabel("y")
    plt.scatter(x, y)
    plt.legend()
    train_rmse = np.sqrt(np.mean((y_pred - y)**2))
    highest_coef = np.max([np.max(clf.coef_), clf.intercept_])
    sqrt_sum_squares_coef = np.sqrt(np.sum(clf.coef_**2) + clf.intercept_**2)
    plt.title('RMSE: {:.2f}\n Highest Coef: {:.2f}\n Sqrt Sum of Squares of Coef: {:.2f}'.format(train_rmse, highest_coef, sqrt_sum_squares_coef))
    #format_axes(plt.gca())
    #plt.savefig('lin_{}.pdf'.format(i+2), transparent=True, bbox_inches="tight")
```

```python
def cost(theta_0, theta_1, x, y):
    s = 0
    for i in range(len(x)):
        y_i_hat = x[i]*theta_1 + theta_0
        s += (y[i]-y_i_hat)**2
    return s/len(x)

x_grid, y_grid = np.mgrid[-4:15:.2, -4:15:.2]

cost_matrix = np.zeros_like(x_grid)
for i in range(x_grid.shape[0]):
    for j in range(x_grid.shape[1]):
        cost_matrix[i, j] = cost(x_grid[i, j], y_grid[i, j], data['x'], data['y'])
```

## visualisation of Bias-Variance Trade-off in Ridge Regression

This animation illustrates how **Ridge Regression (L2 Regularization)** controls model complexity through the **regularization strength (alpha or  λ)**.

- For **small alpha values**, the model has **low bias** but **high variance**, fitting noise and leading to overfitting.
- As **alpha increases**, Ridge penalizes large coefficients, simplifying the model, thus reducing variance but increasing bias.



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
from matplotlib.animation import FuncAnimation, PillowWriter
import os
from IPython.display import Image, display

# Generate synthetic data
x = np.linspace(1, 10, 100)
np.random.seed(10)
y_true = 2.5 * x + 5
y = y_true + np.random.normal(0, 4, size=x.shape)

max_deg = 5
data_x = [x**(i+1) for i in range(max_deg)] + [y]
data_c = ['x'] + ['x_{}'.format(i+1) for i in range(1, max_deg)] + ['y']
data = pd.DataFrame(np.column_stack(data_x), columns=data_c)
data["ones"] = 1

predictors = ['ones', 'x'] + ['x_{}'.format(i+1) for i in range(1, max_deg)]

alphas = np.logspace(-3, 5, 40)
fig, ax = plt.subplots()
line, = ax.plot([], [], 'k', label='Prediction')
scatter = ax.scatter(data['x'], data['y'], label='Train')
true_line, = ax.plot(data['x'], y_true, 'g.', label='True Function')
ax.legend()
ax.set_ylim([min(y)-10, max(y)+10])
ax.set_xlabel("x")
ax.set_ylabel("y")

def init():
    line.set_data([], [])
    return line,

def animate(i):
    alpha = alphas[i]
    regressor = Ridge(alpha=alpha, fit_intercept=False)  # Removed normalize
    regressor.fit(data[predictors], data['y'])
    y_pred = regressor.predict(data[predictors])
    line.set_data(data['x'], y_pred)
    ax.set_title(f"Ridge Regression\nAlpha: {alpha:.2e}")
    return line,

ani = FuncAnimation(fig, animate, frames=len(alphas), init_func=init, blit=True)

# Ensure gif folder exists
gif_dir = r"\\wsl.localhost\Ubuntu\home\suruchi_linux\ml-teaching\notebooks\gifs"
os.makedirs(gif_dir, exist_ok=True)
gif_path = os.path.join(gif_dir, "ridge_animation_bias_variance.gif")

# Save GIF
ani.save(gif_path, writer=PillowWriter(fps=5))
plt.close(fig)

# Display GIF in notebook output
display(Image(filename=gif_path))
```

## visualisation of  Coefficient Shrinkage  in Ridge Regression 

- Unlike Lasso, Ridge Regression **shrinks coefficients continuously** without forcing them to exactly zero.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
from matplotlib.animation import FuncAnimation, PillowWriter
import os
from IPython.display import Image, display

# Generate synthetic data
np.random.seed(42)
x = np.linspace(0, 10, 100)
y_true = 3 * x + 2
y = y_true + np.random.normal(0, 5, size=x.shape)

max_deg = 4
data_x = [x**(i+1) for i in range(max_deg)] + [y]
data_c = ['x'] + [f'x_{i+1}' for i in range(1, max_deg)] + ['y']
data = pd.DataFrame(np.column_stack(data_x), columns=data_c)
data["ones"] = 1
predictors = ['ones', 'x'] + [f'x_{i+1}' for i in range(1, max_deg)]

# Increase number of alphas for smoother animation
alphas = np.logspace(-3, 5, 100)
coef_matrix = []
for alpha in alphas:
    reg = Ridge(alpha=alpha, fit_intercept=False)
    reg.fit(data[predictors], data['y'])
    coef_matrix.append(reg.coef_)
coef_matrix = np.array(coef_matrix)

fig, ax = plt.subplots(figsize=(8, 6))
lines = []
for i in range(coef_matrix.shape[1]):
    (line,) = ax.plot([], [], label=f"$\\theta_{i}$")
    lines.append(line)

ax.set_xscale('log')
ax.set_xlabel("Alpha (Regularization Strength)")
ax.set_ylabel("Coefficient Value")
ax.set_title("Ridge Coefficient Shrinkage Path")
ax.legend()
ax.set_xlim([alphas.min(), alphas.max()])
ax.set_ylim([np.min(coef_matrix)*1.1, np.max(coef_matrix)*1.1])

def init():
    for line in lines:
        line.set_data([], [])
    return lines

def animate(i):
    for j, line in enumerate(lines):
        line.set_data(alphas[:i+1], coef_matrix[:i+1, j])
    ax.set_title(f"Ridge Coefficient Shrinkage\nAlpha: {alphas[i]:.2e}")
    return lines

ani_coef = FuncAnimation(fig, animate, frames=len(alphas), init_func=init, blit=True)

# Save GIF with lower FPS to slow down
gif_dir = r"\\wsl.localhost\Ubuntu\home\suruchi_linux\ml-teaching\notebooks\gifs"
os.makedirs(gif_dir, exist_ok=True)
coef_gif_path = os.path.join(gif_dir, "ridge_coef_shrinkage.gif")
ani_coef.save(coef_gif_path, writer=PillowWriter(fps=6))  # Slower animation with fps=3
plt.close(fig)

# Display GIF in notebook output
display(Image(filename=coef_gif_path))
```

### Ridge Regression (L2 Constraint Visualization)

In Ridge Regression  The solution is the point where the elliptical cost function contours just touch the circular constraint boundary.  This animation shows how Ridge Regression shrinks coefficients within an **L2 circular boundary**. 
As the circle tightens, coefficients are **pulled towards zero** but **never exactly zero out**.
Ridge penalizes large coefficients but keeps **all features active**. This shows how Ridge leads to smooth coefficient shrinkage.





```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.animation import FuncAnimation, PillowWriter
import os
from IPython.display import Image, display

# Synthetic cost function (Least Squares)
def cost(theta_0, theta_1, x, y):
    y_hat = theta_0 + theta_1 * x
    return np.mean((y - y_hat) ** 2)

# Generate synthetic data
np.random.seed(0)
x = np.linspace(0, 5, 20)
y = 2 * x + 3 + np.random.normal(0, 1, size=x.shape)

# Grid for contour plot
theta_0 = np.linspace(-4, 8, 100)
theta_1 = np.linspace(-4, 8, 100)
T0, T1 = np.meshgrid(theta_0, theta_1)
cost_matrix = np.zeros_like(T0)
for i in range(T0.shape[0]):
    for j in range(T0.shape[1]):
        cost_matrix[i, j] = cost(T0[i, j], T1[i, j], x, y)

# OLS solution (unregularized)
X = np.vstack([np.ones_like(x), x]).T
theta_ols = np.linalg.lstsq(X, y, rcond=None)[0]

# Animation setup
radii = np.linspace(7, 0.5, 200)  # Smooth transition with 200 frames
fig, ax = plt.subplots(figsize=(6, 6))

def animate(i):
    ax.clear()
    radius = radii[i]
    # Cost Contours (Ellipses)
    levels = np.percentile(cost_matrix, [10, 20, 40, 60, 80])
    ax.contour(T0, T1, cost_matrix, levels=levels, colors='black')
    # L2 Constraint Circle
    circle = Circle((0, 0), radius, color='cyan', alpha=0.3, label=f'L2 constraint ($\\theta_0^2+\\theta_1^2 \\leq$ {radius**2:.2f})')
    ax.add_patch(circle)
    # Axes lines
    ax.axhline(0, color='gray', lw=1)
    ax.axvline(0, color='gray', lw=1)
    # OLS Solution
    ax.plot(theta_ols[0], theta_ols[1], marker='*', color='blue', markersize=14, label='OLS Solution')
    # L2 Boundary (Circle Outline)
    theta = np.linspace(0, 2 * np.pi, 100)
    ax.plot(radius * np.cos(theta), radius * np.sin(theta), color='navy', lw=2)
    # Ridge Solution Projection
    coef_norm = np.sqrt(theta_ols[0]**2 + theta_ols[1]**2)
    if coef_norm > radius:
        scale = radius / coef_norm
        ridge_solution = scale * theta_ols
    else:
        ridge_solution = theta_ols
    ax.plot(ridge_solution[0], ridge_solution[1], 'ro', label='Ridge Solution')
    ax.annotate(f"({ridge_solution[0]:.2f}, {ridge_solution[1]:.2f})", 
                (ridge_solution[0], ridge_solution[1]), textcoords="offset points", xytext=(10, -10), ha='center')
    ax.set_xlim(-4, 8)
    ax.set_ylim(-4, 8)
    ax.set_xlabel(r'$\theta_0$')
    ax.set_ylabel(r'$\theta_1$')
    ax.set_title(f'Ridge Regression Geometry\nL2 Constraint Radius: {radius:.2f}')
    ax.legend(loc='upper right')
    ax.set_aspect('equal')
    return ax.patches + ax.lines

# Save GIF to desired location
gif_dir = r"\\wsl.localhost\Ubuntu\home\suruchi_linux\ml-teaching\notebooks\gifs"
os.makedirs(gif_dir, exist_ok=True)
gif_path = os.path.join(gif_dir, "ridge_l2_constraint_fixed.gif")

ani = FuncAnimation(fig, animate, frames=len(radii), blit=True)
ani.save(gif_path, writer=PillowWriter(fps=8))
plt.close(fig)

# Display in Notebook
display(Image(filename=gif_path))
```

```python
#Define input array with angles from 60deg to 300deg converted to radians
x = np.array([i*np.pi/180 for i in range(60,300,4)])
np.random.seed(10)  #Setting seed for reproducability
y = 4*x + 7 + np.random.normal(0,3,len(x))

y_true = 4*x + 7
max_deg = 20

data_x = [x**(i+1) for i in range(max_deg)] + [y]
data_c = ['x'] + ['x_{}'.format(i+1) for i in range(1,max_deg)] + ['y']
data = pd.DataFrame(np.column_stack(data_x),columns=data_c)
data["ones"] = 1
plt.plot(data['x'],data['y'],'.', label='Data Points')
plt.plot(data['x'], y_true,'g', label='True Function')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.savefig('true_function.pdf', transparent=True, bbox_inches="tight")
```

```python
def cost(theta_0, theta_1, x, y):
    s = 0
    for i in range(len(x)):
        y_i_hat = x[i]*theta_1 + theta_0
        s += (y[i]-y_i_hat)**2
    return s/len(x)

x_grid, y_grid = np.mgrid[-4:15:.2, -4:15:.2]

cost_matrix = np.zeros_like(x_grid)
for i in range(x_grid.shape[0]):
    for j in range(x_grid.shape[1]):
        cost_matrix[i, j] = cost(x_grid[i, j], y_grid[i, j], data['x'], data['y'])
```

```python
def cost_lasso(theta_0, theta_1, x, y, lamb):
    s = 0
    for i in range(len(x)):
        y_i_hat = x[i]*theta_1 + theta_0
        s += (y[i]-y_i_hat)**2 + lamb*(abs(theta_0) + abs(theta_1))
    return s/len(x)

x_grid, y_grid = np.mgrid[-4:15:.2, -4:15:.2]

```

```python
#lambda  = 1000 cost curve tends to lasso objective. 
fig = plt.figure(figsize=(7,7))
lamb_list = [10,100,1000]
for lamb in lamb_list:
    lasso_cost_matrix = np.zeros_like(x_grid)
    for i in range(x_grid.shape[0]):
        for j in range(x_grid.shape[1]):
            lasso_cost_matrix[i, j] = cost_lasso(x_grid[i, j], y_grid[i, j], data['x'], data['y'],lamb)


    ax = plt.axes(projection='3d')
    ax.plot_surface(x_grid, y_grid, lasso_cost_matrix,cmap='viridis', edgecolor='none')

    ax.set_title('Least squares objective function');
    ax.set_xlabel(r"$\theta_0$")
    ax.set_ylabel(r"$\theta_1$")
    ax.set_xlim([-4,15])
    ax.set_ylim([-4,15])

    u = np.linspace(0, np.pi, 30)
    v = np.linspace(0, 2 * np.pi, 30)

    # x = np.outer(500*np.sin(u), np.sin(v))
    # y = np.outer(500*np.sin(u), np.cos(v))
    # z = np.outer(500*np.cos(u), np.ones_like(v))
    # ax.plot_wireframe(x, y, z)

    ax.view_init(45, 120)
    plt.savefig('lasso_lamb_{}_surface.pdf'.format(lamb), transparent=True, bbox_inches="tight")
```

## visualisation of  Coefficient Shrinkage  in Lasso Regression 

- Lasso Regression **shrinks coefficients continuously** to exactly zero that  helps in feature extraction .

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
from matplotlib.animation import FuncAnimation, PillowWriter
import os
from IPython.display import Image, display

# Generate synthetic data
np.random.seed(42)
x = np.linspace(0, 10, 100)
y_true = 3 * x + 2
y = y_true + np.random.normal(0, 5, size=x.shape)

max_deg = 4
data_x = [x**(i+1) for i in range(max_deg)] + [y]
data_c = ['x'] + [f'x_{i+1}' for i in range(1, max_deg)] + ['y']
data = pd.DataFrame(np.column_stack(data_x), columns=data_c)
data["ones"] = 1
predictors = ['ones', 'x'] + [f'x_{i+1}' for i in range(1, max_deg)]

# Scale features for better convergence
scaler = StandardScaler()
X_scaled = scaler.fit_transform(data[predictors])

# Increase alpha range and number of frames for full shrinkage
alphas = np.logspace(-3, 4, 120)  # More frames, larger max alpha
coef_matrix = []
for alpha in alphas:
    reg = Lasso(alpha=alpha, fit_intercept=False, max_iter=100000)
    reg.fit(X_scaled, data['y'])
    coef_matrix.append(reg.coef_)
coef_matrix = np.array(coef_matrix)

fig, ax = plt.subplots(figsize=(7, 5))
lines = []
for i in range(coef_matrix.shape[1]):
    (line,) = ax.plot([], [], label=f"$\\theta_{i}$")
    lines.append(line)
ax.set_xscale('log')
ax.set_xlabel("Alpha (Regularization Strength)")
ax.set_ylabel("Coefficient Value")
ax.set_title("Lasso Regularization Path")
ax.legend()
ax.set_xlim([alphas.min(), alphas.max()])
ax.set_ylim([np.min(coef_matrix)*1.1, np.max(coef_matrix)*1.1])

def init():
    for line in lines:
        line.set_data([], [])
    return lines

def animate(i):
    for j, line in enumerate(lines):
        line.set_data(alphas[:i+1], coef_matrix[:i+1, j])
    ax.set_title(f"Lasso Path: Coefficient Shrinkage\nAlpha: {alphas[i]:.2e}")
    return lines

ani = FuncAnimation(fig, animate, frames=len(alphas), init_func=init, blit=True)

gif_dir = r"\\wsl.localhost\Ubuntu\home\suruchi_linux\ml-teaching\notebooks\gifs"
os.makedirs(gif_dir, exist_ok=True)
gif_path = os.path.join(gif_dir, "lasso_regularization_path.gif")
ani.save(gif_path, writer=PillowWriter(fps=10))  # Faster animation for more frames
plt.close(fig)

display(Image(filename=gif_path))
```

### visualisation showing how  **LASSO induces sparsity**

This animation illustrates how **LASSO (L1 Regularization)** constrains the parameter space using a **diamond-shaped constraint region** (|θ₀| + |θ₁| ≤ constant).

- The **black contours** represent the **Least Squares Cost Function**.
- The **orange diamond** is the **L1 constraint boundary**.
- The **OLS solution (blue star)** shows the optimal unregularized parameters.
- As the **diamond shrinks**, the feasible parameter space reduces.
- **LASSO tends to "clip" coefficients at the diamond's corners**, pushing some θ values to exactly **zero**.



```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
from matplotlib.animation import FuncAnimation, PillowWriter
import os
from IPython.display import Image, display

# Synthetic cost function (least squares)
def cost(theta_0, theta_1, x, y):
    y_hat = theta_0 + theta_1 * x
    return np.mean((y - y_hat) ** 2)

# Generate synthetic data
np.random.seed(0)
x = np.linspace(0, 5, 20)
y = 2 * x + 3 + np.random.normal(0, 1, size=x.shape)

# Grid for contours
theta_0 = np.linspace(-4, 8, 100)
theta_1 = np.linspace(-4, 8, 100)
T0, T1 = np.meshgrid(theta_0, theta_1)
cost_matrix = np.zeros_like(T0)
for i in range(T0.shape[0]):
    for j in range(T0.shape[1]):
        cost_matrix[i, j] = cost(T0[i, j], T1[i, j], x, y)

# L1 diamond vertices generator
def l1_diamond_vertices(radius):
    return np.array([
        [0, radius],
        [radius, 0],
        [0, -radius],
        [-radius, 0]
    ])

# Find OLS solution (unregularized)
X = np.vstack([np.ones_like(x), x]).T
theta_ols = np.linalg.lstsq(X, y, rcond=None)[0]

# Animation setup
radii = np.linspace(7, 0, 100)  # More frames, shrink diamond to zero
fig, ax = plt.subplots(figsize=(6, 6))

def animate(i):
    ax.clear()
    # Draw cost contours
    levels = np.percentile(cost_matrix, [10, 20, 40, 60, 80])
    ax.contour(T0, T1, cost_matrix, levels=levels, colors='black')
    # Draw L1 diamond
    radius = radii[i]
    diamond = Polygon(l1_diamond_vertices(radius), closed=True, color='orange', alpha=0.3, label=f'L1 constraint (|θ₀|+|θ₁|≤{radius:.2f})')
    ax.add_patch(diamond)
    # Axes
    ax.axhline(0, color='gray', lw=1)
    ax.axvline(0, color='gray', lw=1)
    # OLS solution
    ax.plot(theta_ols[0], theta_ols[1], marker='*', color='blue', markersize=14, label='OLS Solution')
    # Show diamond corners
    corners = l1_diamond_vertices(radius)
    ax.scatter(corners[:,0], corners[:,1], color='red', s=40, label='Diamond Corners')
    ax.set_xlim(-4, 8)
    ax.set_ylim(-4, 8)
    ax.set_xlabel(r'$\theta_0$')
    ax.set_ylabel(r'$\theta_1$')
    ax.set_title(f'L1 Constraint Geometry\nDiamond radius: {radius:.2f}')
    ax.legend(loc='upper right')
    ax.set_aspect('equal')
    return ax.patches + ax.lines

gif_dir = r"\\wsl.localhost\Ubuntu\home\suruchi_linux\ml-teaching\notebooks\gifs"
os.makedirs(gif_dir, exist_ok=True)
gif_path = os.path.join(gif_dir, "lasso_l1_geometry.gif")

ani = FuncAnimation(fig, animate, frames=len(radii), blit=True)
ani.save(gif_path, writer=PillowWriter(fps=10))  # Smoother animation
plt.close(fig)
display(Image(filename=gif_path))
```
